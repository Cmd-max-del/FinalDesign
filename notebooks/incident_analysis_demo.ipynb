# Incident Data Analysis Demo
Use this notebook to perform exploratory analysis on police incident records.

```python
import csv
from collections import Counter
from pathlib import Path
```

```python
csv_path = Path('..') / 'data' / 'sample' / 'incidents_sample.csv'
with open(csv_path, 'r', encoding='utf-8') as f:
    rows = list(csv.DictReader(f))
rows[:5]
```

```python
durations = [int(r['handling_duration_min']) for r in rows]
units = [int(r['police_unit_count']) for r in rows]
summary = {
    'total_incidents': len(rows),
    'avg_duration_min': round(sum(durations) / len(durations), 2),
    'avg_police_units': round(sum(units) / len(units), 2),
}
summary
```

```python
type_distribution = Counter(r['incident_type'] for r in rows)
sorted(type_distribution.items(), key=lambda x: x[1], reverse=True)
```
